```python
import pandas as pd

# Original-LLM-Antworten einlesen und überprüfen
answers = pd.read_excel("../original/model_opinions2021.xlsx")
display(answers)

## Antworten in Zahlen überführen (0 für nein, 1 für neutral, 2 für ja, wie in viz.ipynb)
answers_clean = answers.replace(to_replace=r'.*Nein.*', value=0, regex=True)
answers_clean = answers_clean.replace(to_replace=r'.*No.*', value=0, regex=True)
answers_clean = answers_clean.replace(to_replace=r'.*Neutral.*', value=1, regex=True)
answers_clean = answers_clean.replace(to_replace=r'.*Ja.*', value=2, regex=True)
answers_clean = answers_clean.replace(to_replace=r'.*Yes.*', value=2, regex=True)
display(answers_clean)
```

```python
# Wahl-O-Mat Datensatz einlesen
party_raw = pd.read_excel("../external_data/wahlomat_europa_2024.xlsx",  sheet_name="Datensatz EU 2024")
print(party_raw["Position: Position"].value_counts())

# Positionen in Zahlen umwandeln und relevante Variablen filtern
party_clean = party_raw[[
    "Partei: Nr.",
    "Partei: Kurzbezeichnung",
    "These: Nr.",
    "These: Titel",
    "Position: Position"]].copy()

party_clean["Position: Position"] = party_clean["Position: Position"].replace({
    "stimme nicht zu": 0,
    "neutral": 1,
    "stimme zu": 2})

display(party_clean.head())
```

```python
# Vergleich der Datensätze - sind die Fragen in derselben Reihenfolge?
questions = pd.read_json("../original/theses_2021.json") # Thesenfile einladen, mit dem Modelantworten generiert wurden
wahlomat = party_clean.drop_duplicates("These: Nr.").sort_values("These: Nr.") # aus dem Wahl-O-Mat Doc auch nur die Thesen in ihrer Reihenfolge extrahierenb

titel_wom = wahlomat["These: Titel"].tolist()
titel_llm = questions["Description_DE"].tolist()

print("Gleiche Reihenfolge:", titel_wom == titel_llm)
```

```python
# Parteiantworten und LLM-Antworten vergleichen:
## Parteiantworten in dasselbe Format umwandeln:
party_wide = party_clean.pivot(
    index="Partei: Kurzbezeichnung",
    columns="These: Nr.",
    values="Position: Position")
party_wide.columns = party_wide.columns.astype(str)

## Funktion zur Berechnung der Alignment-Scores:
fragen = [str(i) for i in range(1, 39)]

def alignment_score(model, partei):
    llm = answers_clean.loc[model, fragen]
    wom = party_wide.loc[partei, fragen]

    score = (2 - abs(llm - wom)).sum() / 76 * 100
    return round(score, 1)

## Scores berechnen:
## Scores berechnen:
results_german = []
results_english = []

for partei in party_wide.index:
    row_german = {"Political Party": partei}
    row_english = {"Political Party": partei}

    for i in answers_clean.index:
        model = answers_clean.loc[i, "model"]
        language = answers_clean.loc[i, "language"]

        if language == "german":
            row_german[model] = alignment_score(i, partei)
        else:
            row_english[model] = alignment_score(i, partei)

    results_german.append(row_german)
    results_english.append(row_english)

scores_german = pd.DataFrame(results_german)
scores_english = pd.DataFrame(results_english)
display(scores_german)
```
